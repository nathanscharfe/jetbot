# OLED Hello Example

This notebook writes a simple message to the JetBot's OLED display.

Notes:
- This version is aimed at the SparkFun JetBot kit, which ships with a Qwiic Micro OLED.
- The default stats screen may overwrite your message while `jetbot_stats.service` is running.
- If that happens, open a terminal in Jupyter and run `sudo systemctl stop jetbot_stats.service`.
- To restore the default OLED behavior later, run `sudo systemctl start jetbot_stats.service`.


````python
import subprocess


def service_state(name: str) -> str:
    result = subprocess.run(
        ["systemctl", "is-active", name],
        check=False,
        capture_output=True,
        text=True,
    )
    return (result.stdout or result.stderr).strip() or "unknown"


state = service_state("jetbot_stats.service")
print(f"jetbot_stats.service: {state}")

if state == "active":
    print("The default stats service is active and may overwrite the OLED.")
    print("If needed, stop it in a Jupyter terminal with: sudo systemctl stop jetbot_stats.service")
````

````python
import qwiic_micro_oled


def centered_x(oled, text: str) -> int:
    char_width = oled.get_font_width()
    return max(0, (oled.get_lcd_width() - len(text) * char_width) // 2)


def write_hello(lines=("Hello", "Nathan")):
    oled = qwiic_micro_oled.QwiicMicroOled()

    if not oled.is_connected():
        raise RuntimeError(
            "Qwiic Micro OLED was not detected. Check the Qwiic cable and confirm the SparkFun OLED is attached."
        )

    oled.begin()
    oled.clear(oled.PAGE)
    oled.set_font_type(0)

    first_y = 10
    line_gap = oled.get_font_height() + 4

    for index, line in enumerate(lines):
        oled.set_cursor(centered_x(oled, line), first_y + index * line_gap)
        oled.print(line)

    oled.display()
    return oled


oled = write_hello()
print('Wrote "Hello Nathan" to the OLED.')
````

To go back to the default OLED status screen, start the service again in a terminal:

```bash
sudo systemctl start jetbot_stats.service
```
